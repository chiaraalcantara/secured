```python
from pytubefix import YouTube
from pytubefix.cli import on_progress
# https://pytubefix.readthedocs.io/

url = "https://www.youtube.com/watch?v=2lAe1cqCOXo"

yt = YouTube(url, on_progress_callback=on_progress)
print(yt.title)  # Print video title
ys = yt.streams.filter(res="360p", file_extension="mp4").first()
ys.download()
```

```python
from pytubefix import Search

keywords = ["classroom lecture", "student engagement", "classroom interaction"]
for word in keywords: 
    results = Search(word)
    for video in results.videos[:20]:
        print(f'Title: {video.title}')
        print(f'URL: {video.watch_url}')
        print(f'Duration: {video.length/60} mins')
        print('---')

# results.get_next_results()
```
